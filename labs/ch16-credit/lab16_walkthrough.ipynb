# Lab 16: Do Credit Spreads Compensate Mainly for Expected Default Losses?

*Financial Engineering and the Evidence*, Chapter 16.

**Claim on trial:** the yield spread of corporate bonds over Treasuries mostly reflects the losses investors expect from default.

You will measure Moody's Baa and Aaa spreads since 1919, set your own assumptions for default rates and recoveries, compute how much of the spread expected losses can explain, and ask whether the rest behaves like a risk premium. Before each step, write down a prediction.

```python
from pathlib import Path
from IPython.display import Image, display
import lab16

SOURCE = "fred"       # "fred" downloads AAA, BAA, GS10, USREC, INDPRO; "sim" uses fake data to test the code
FILES = {"aaa": None, "baa": None, "gs10": None, "usrec": None, "indpro": None,
         "ebp": None}  # optional: Gilchrist-Zakrajsek ebp_csv.csv from the Federal Reserve
PD_BAA, PD_AAA, RECOVERY = 0.0027, 0.0, 0.38   # Moody's long-run averages; change them and rerun
OUT = Path("report")
OUT.mkdir(exist_ok=True)

d = lab16.load(SOURCE, FILES)
SIM = d["simulated"]
s = lab16.spreads(d["df"])
print(s.index[0], "to", s.index[-1], "| simulated:", SIM)
```

## Part A: Spreads

**Predict first:** what is the average spread of Baa bonds over 10-year Treasuries since 1953? How much wider is it in recessions?

```python
a = lab16.part_a(s)
lab16.fig_spreads(s, OUT, SIM)
display(Image(OUT / "fig16_1_spreads.png"))
a["table"].round(4)
```

## Part B: Expected losses versus spreads

**Predict first:** a Baa bond defaults with probability about 0.27% a year and recovers about 38% of face value. What spread would exactly compensate for expected losses? What share of the average spread is that?

```python
b = lab16.part_b(s, PD_BAA, PD_AAA, RECOVERY)
lab16.fig_decomposition(b, OUT, SIM)
lab16.fig_breakeven(b, s, OUT, SIM)
display(Image(OUT / "fig16_2_decomposition.png"))
display(Image(OUT / "fig16_3_breakeven.png"))
b["table"].round(4)
```

## Part C: The residual as a risk premium

**Predict first:** when the Baa–Aaa spread is in its top fifth, what happens to industrial production over the next year?

```python
c = lab16.part_c(s, d["df"], d["ebp"])
lab16.fig_forecast(c, OUT, SIM)
display(Image(OUT / "fig16_4_cycle.png"))
c["forecast"].round(3)
```

## Part D: Stress tests

**Predict first:** what default rate would be needed for expected losses to explain the whole Baa–Treasury spread? Has the Baa default rate ever been that high for a sustained period?
