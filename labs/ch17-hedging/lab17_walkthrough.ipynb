# Lab 17: Does Delta Hedging Remove Option Risk?

*Financial Engineering and the Evidence*, Chapter 17.

**Claim on trial:** an option seller who delta-hedges as Black–Scholes prescribes is left with little risk.

Each month since 1990 you will sell a one-month at-the-money call at its Black–Scholes price, using the VIX as the implied volatility, and hedge it along the market's actual daily path. Then you will do the same on simulated Black–Scholes paths, where the theory holds exactly, and compare. Before each step, write down a prediction.

```python
from pathlib import Path
from IPython.display import Image, display
import lab17

SOURCE = "french"     # "french" downloads daily factors and the VIX; "sim" uses fake data to test the code
FILES = {"ff3": None, "vix": None}   # or paths to files you saved in data/
OUT = Path("report")
OUT.mkdir(exist_ok=True)

d = lab17.load(SOURCE, FILES)
SIM = d["simulated"]
months = lab17.month_paths(d)
print(len(months), "months from", months[0]["month"], "| simulated:", SIM)
```

## Part A: How much risk is left?

**Predict first:** with daily hedging on Black–Scholes paths, the hedging error has a standard deviation of roughly √(π/4)·σ·vega/√N, about a fifth of the premium for a one-month option. Will real paths leave more or less?

```python
res = lab17.run_hedges(months, n_sims=20)   # about a minute
a = lab17.part_a(res)
lab17.fig_hist(res, OUT, SIM)
display(Image(OUT / "fig17_1_hist.png"))
a[["paths", "hedging", "mean", "sd", "skewness", "q01", "worst", "risk_removed"]].round(3)
```
