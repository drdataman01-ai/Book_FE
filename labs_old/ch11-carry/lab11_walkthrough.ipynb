# Lab 11: Do High-Interest Currencies Depreciate Enough to Offset Their Yield?

*Financial Engineering and the Evidence*, Chapter 11.

**Claim on trial (uncovered interest parity):** the currency of a high-interest country is expected to fall by its interest advantage, so borrowing low and lending high earns nothing on average.

You will run Fama regressions for nine currencies against the dollar, build a G10 carry portfolio, and look at when and how it loses money. Before each step, write down a prediction.

```python
from pathlib import Path
from IPython.display import Image, display
import lab11
from fintest.fx import excess_returns

SOURCE = "fred"       # "fred" downloads exchange rates and interest rates; "sim" uses fake data to test the code
FILES = {}            # optional: {"EXJPUS": "data/EXJPUS.csv", ...} for files saved from FRED
OUT = Path("report")
OUT.mkdir(exist_ok=True)

d = lab11.load(SOURCE, FILES)
SIM = d["simulated"]
er = excess_returns(d["spot"], d["rate"], d["us"])
print(list(d["spot"].columns), "| simulated:", SIM, "| missing:", d.get("missing"))
```

## Part A: Fama regressions

**Predict first:** UIP says the slope is 1: a currency with a 3% interest advantage over the dollar should fall 3% a year. What slopes will you find?

```python
a = lab11.part_a(er)
lab11.fig_fama(a, OUT, SIM)
display(Image(OUT / "fig11_1_fama.png"))
a.round(2)
```

## Part B: The carry portfolio

**Predict first:** going long the three highest-rate currencies and short the three lowest, how much of the interest differential will exchange-rate moves take back?

```python
b = lab11.part_b(er)
lab11.fig_cumulative(b, OUT, SIM)
lab11.fig_hist(b, OUT, SIM)
display(Image(OUT / "fig11_2_cumulative.png"))
display(Image(OUT / "fig11_3_hist.png"))
print(b["components"])
b["table"].round(3)
```

## Part C: Stress tests

**Predict first:** in which months will carry lose most? What will the stock market and the VIX be doing?

```python
c = lab11.part_c(er, d, b)
display(c["subperiods"].round(3))
display(c["costs"].round(4))
display(c["exposures"].round(3))
c["worst"].round(3)
```

## Part D: The investor's view

```python
dd = lab11.part_d(b)
lab11.fig_drawdown(dd, OUT, SIM)
lab11.fig_vix(b, d, OUT, SIM)
display(Image(OUT / "fig11_4_drawdown.png"))
display(Image(OUT / "fig11_5_vix.png"))
{k: v for k, v in dd.items() if k != "drawdown"}
```
